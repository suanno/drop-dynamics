# Verify if the coupling with the film in the $v_1$ formula is negligible for large droplets

We consider the formula
$$\left( \int \frac{(\hat{h}_{0}^{in})^2}{Q_{0}^{in}}d\xi\right)\mathbf{v_{1}}=\left( -\nabla_{\chi}\int \hat{W}_{0}^{in}d\xi \right)+\left(\partial_{\chi}\partial_{h}W_{0}^{out}\int \hat{h}_{0}^{in}\frac{Q_{0}^{out}}{Q_{0}^{in}}d\xi \right)$$
where
$$Q=h^3/3\eta$$
$$W=A(\chi)w(h)-ch\chi;\quad \chi = \epsilon x$$
$$w(h)=h_a^3\frac15 h^{-5}-\frac12 h^{-2}$$
Then we can define
$$K= \int \frac{(\hat{h}_{0}^{in})^2}{(h_{0}^{in})^3}d\xi;\quad \hat{\Omega}=\int d\xi \hat{w}_{0}^{in};\quad m=\int d\xi \hat{h}_{0}^{in};\quad I=(h_{0}^{out})^3\int d\xi\frac{ \hat{h}_{0}^{in}}{(h_{0}^{in})^3}$$
and rewrite the formula as

$$(3\eta K) v_{1}= \left(-\nabla_{\chi}A\hat{\Omega} + c m \right)+ \left(  \nabla_{\chi}A w'_{out}I-cI \right)$$


# What changes if $\partial_{\chi}\partial_{h}W_0^{out}\to\nabla_{\chi}\partial_{h}W_0^{out}$?

$$\nabla_{\chi}\partial_{h}W_0^{out} = \partial_{\chi}\partial_{h}W_0^{out}+\partial_{hh}W_0^{out}\nabla_{\chi}h_0^{out}$$
so the additional term if $W=A(\chi)w(h);\quad \chi = \epsilon x$ would be
$$\partial_{hh}W_0^{out}\nabla_{\chi}h_0^{out}=w''_{out}A(\chi_*)\nabla_{\chi}h_0^{out}$$

```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
# Continuation to compute h0in for different values of hmax or h0out
data = np.loadtxt(
    "1D_c0_continuation_normalized.txt",
    delimiter=',',
    unpack=True
)
numrows = len(data[0,:])
K = data[1:, numrows-1]
I = data[1:, numrows-2]
Omega = data[1:, numrows-3]
m = data[1:, numrows-4]
h0out_cont = data[1:, numrows-5]
hmax_cont = data[1:, numrows-6]
L = 25 # Size of the system
c0 = data[1:, numrows-7]   # concentration times size of the system gives mass
ha = np.mean(data[1:, numrows-8])   # It is a constant
def dwetting_pot(h):
        return (- h**(-6) + h**(-3))
def ddwetting_pot(h):
        return (6*h**(-7) - 3*h**(-4))
```

```python
plt.plot(c0, h0out_cont)
plt.title("Continuation in $c=L'^{-1}\\int H_0^{\\mathrm{in}}dx'$ (increasing)\n On the left we see the unstable part of the drop branch")
plt.xlabel('$c$')
plt.ylabel('$H_0^{\\mathrm{out}}$')
```

```python
plt.plot(h0out_cont, m, label='m')
plt.plot(h0out_cont, I, label='I')
plt.ylim(top=100)
plt.xlabel('$H_0^{\\mathrm{out}}$')
plt.title("The drop is large if $H_0^{\\mathrm{out}}\\to 1$\nm > I")
```

```python
plt.plot(h0out_cont, h0out_cont*ddwetting_pot(h0out_cont))
plt.plot(h0out_cont, dwetting_pot(h0out_cont))
plt.title('$H_0^{\mathrm{out}} \partial_{HH}\omega^{\mathrm{out}} > \partial_{\mathrm{H}}\omega^{\mathrm{out}} $')
plt.xlabel('$H_0^{\\mathrm{out}}$')
```

```python
plt.plot(h0out_cont, Omega)
plt.plot(h0out_cont, I*h0out_cont*ddwetting_pot(h0out_cont))
plt.title("$\Omega > I\cdot H_0^{\mathrm{out}}\partial_{HH}\omega^{\mathrm{out}}$ if $H_0^{\mathrm{out}}<H^*$\n (sufficiently large drop)")

plt.xlabel('$H_0^{\\mathrm{out}}$')
```
